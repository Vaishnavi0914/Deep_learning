```python
import numpy as np
import matplotlib.pyplot as plt

# Input and output
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,0,0,1])

# Initialize parameters
w1 = 1.2
w2 = 0.6
threshold = 1
lr = 0.5
epochs = 10

# Training
for epoch in range(epochs):

    print("\nEpoch:", epoch+1)
    print("Initial Weights:", w1, w2)

    error_count = 0

    for i in range(len(X)):
        x1, x2 = X[i]

        net = x1*w1 + x2*w2

        # Activation function
        pred = 1 if net >= threshold else 0

        # Calculate error
        error = y[i] - pred

        if error != 0:
            error_count += 1

            w1 = w1 + lr*error*x1
            w2 = w2 + lr*error*x2

        print("Input:", X[i], "Actual:", y[i],
              "Predicted:", pred, "Error:", error)
        print("Updated Weights:", w1, w2)

    print("Total Errors:", error_count)
    print("Final Weights:", w1, w2)

    if error_count == 0:
        print("Perceptron Converged!")
        break

# Testing
print("\nTesting Final Model:")

for i in range(len(X)):
    x1, x2 = X[i]

    net = x1*w1 + x2*w2
    pred = 1 if net >= threshold else 0

    print(X[i], "Predicted:", pred, "Actual:", y[i])

# Decision boundary
x = np.array([-0.5, 1.5])
y_line = (threshold - w1*x) / w2

plt.scatter(X[:,0], X[:,1], c=y, cmap="bwr")
plt.plot(x, y_line, 'k-', label="Decision Boundary")

plt.xlabel("X1")
plt.ylabel("X2")
plt.title("AND Gate Perceptron")
plt.legend()
plt.grid()
plt.show()
```
